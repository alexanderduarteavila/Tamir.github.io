```
import os
import json
import hashlib
import random
import sqlite3
from datetime import datetime
from typing import Dict, List, Any, Optional
import time # Import time

# =========================================================
# 🧠 VIDA CORE - SISTEMA DE PENSAMIENTO CUÁDRUPLE EVOLUCIONADO
# Versión 12.3 - MÁXIMA TRAZABILIDAD (Context Manager + Parámetros Nombrados)
# =========================================================

class VidaCoreEvolucionado:
    """
    🚀 SISTEMA DE PENSAMIENTO CUÁDRUPLE EVOLUCIONADO
    Versión 12.3 - Fortificación de Arquitectura de Producción y Trazabilidad.
    """

    # Mapeo para garantizar la extracción completa de insights (R1 Fortificada)
    INSIGHT_MAPPING = {
        "analisis_tecnico": ["patrones_identificados", "recomendaciones_implementacion", "preguntas_clave", "observaciones_tecnicas"],
        "investigacion_detective": ["pistas_detectadas", "riesgos_identificados", "hipotesis_iniciales", "conexiones_ocultas"],
        "exploracion_curiosa": ["preguntas_exploratorias", "horizones_nuevos", "experimentos_sugeridos", "futuros_posibles"],
        "analisis_economico": ["metricas_financieras", "modelo_negocio_sugerido", "analisis_rentabilidad", "oportunidades_mercado"]
    }

    def __init__(self, nombre: str = "VIDA-CORE-EVOLUCIONADO"):
        self.nombre = nombre
        self.creador = "Daniel"
        self.version = "12.3 - Máxima Trazabilidad"
        self.estado = "PENSAMIENTO_ACTIVO"
        self.base_datos = "vida_core_evolucionado.db"
        self.umbral_aprendizaje = 0.7

        # 🎯 ARQUITECTURA CUÁDRUPLE MEJORADA
        self.modalidades_activas = {
            "tecnico": {"nivel": 10, "eficacia": 0.92, "peso_analisis": 1.2, "patrones": ["código", "software", "algoritmo", "sistema", "seguridad", "blockchain", "contrato", "api"]},
            "detective": {"nivel": 10, "eficacia": 0.88, "peso_analisis": 1.1, "patrones": ["investigar", "patrón", "vulnerabilidad", "forense", "auditoría", "transacción", "conexión"]},
            "curiosidad": {"nivel": 10, "eficacia": 0.85, "peso_analisis": 1.0, "patrones": ["qué pasaría si", "innovación", "futuro", "disrupción", "explorar", "nuevo", "optimizar"]},
            "economico": {"nivel": 9, "eficacia": 0.80, "peso_analisis": 0.9, "patrones": ["costo", "inversión", "rentabilidad", "ROI", "mercado", "negocio", "modelo", "financiero"]}
        }

        # 🚀 SISTEMA DE PERSISTENCIA AVANZADO
        self._inicializar_base_datos_avanzada()

        # ⚡ CACHE DE ALTO RENDIMIENTO
        self.cache_patrones = self._construir_cache_patrones()
        self.cache_consultas = {}
        self.max_cache_size = 1000

        # 📊 MÉTRICAS Y ESTADO
        self.metricas = {
            "total_analisis": self._obtener_total_analisis(),
            "eficiencia_promedio": self._obtener_eficiencia_promedio(),
            "patrones_aprendidos": self._obtener_total_patrones(),
            "mejoras_detectadas": 0,
            "cache_hits": 0,
            "cache_misses": 0
        }

        print(f"🚀 {self.nombre} INICIALIZADO - v{self.version}")
        print(f"📊 Base: {self.metricas['total_analisis']} análisis | {self.metricas['patrones_aprendidos']} patrones persistentes.")

    # ------------------------------------------------------------------
    #  Métodos de bajo nivel (DB, caché, métricas...) - CORREGIDOS (C2)
    # ------------------------------------------------------------------

    def _inicializar_base_datos_avanzada(self):
        """Crea tablas e índices si no existen (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()

                # Tabla principal de análisis
                cur.execute('''
                    CREATE TABLE IF NOT EXISTS analisis_profundos (
                        id INTEGER PRIMARY KEY AUTOINCREMENT,
                        timestamp TEXT NOT NULL,
                        consulta_hash TEXT UNIQUE,
                        consulta_texto TEXT NOT NULL,
                        eficacia_analisis REAL,
                        patrones_detectados TEXT
                    )
                ''')

                # Aprendizajes automáticos
                cur.execute('''
                    CREATE TABLE IF NOT EXISTS aprendizajes_automaticos (
                        id INTEGER PRIMARY KEY AUTOINCREMENT,
                        patron TEXT UNIQUE,
                        eficacia_demostrada REAL DEFAULT 0.7,
                        veces_utilizado INTEGER DEFAULT 1,
                        ultimo_uso TEXT,
                        categoria TEXT
                    )
                ''')

                cur.execute('CREATE INDEX IF NOT EXISTS idx_hash ON analisis_profundos(consulta_hash)')
                cur.execute('CREATE INDEX IF NOT EXISTS idx_patron ON aprendizajes_automaticos(patron)')
                conn.commit()
        except Exception as e:
            print(f"❌ ERROR CRÍTICO [INIT_DB]: No se pudo inicializar la base de datos. {e}")

    def _construir_cache_patrones(self) -> Dict[str, set]:
        """Carga patrones fijos y aprendidos en el cache de memoria (C2: Context Manager)."""
        cache = {}
        # Cargar patrones fijos de las modalidades
        for modalidad, datos in self.modalidades_activas.items():
            cache[modalidad] = set(datos['patrones'])

        # Cargar patrones aprendidos desde la DB
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()
                cursor.execute("SELECT patron, categoria FROM aprendizajes_automaticos")
                for patron, categoria in cursor.fetchall():
                    cache.setdefault(categoria.lower(), set()).add(patron)
        except Exception:
            pass # Si falla, solo usamos cache fijo

        return cache

    def _obtener_total_analisis(self) -> int:
        """Obtiene el total de análisis desde la DB (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()
                cur.execute('SELECT COUNT(*) FROM analisis_profundos')
                return cur.fetchone()[0]
        except Exception:
            return 0

    def _obtener_eficiencia_promedio(self) -> float:
        """Calcula eficiencia promedio desde DB (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()
                cur.execute('SELECT AVG(eficacia_analisis) FROM analisis_profundos WHERE eficacia_analisis > 0')
                resultado = cur.fetchone()[0]
                return resultado if resultado else 0.75
        except Exception:
            return 0.75

    def _obtener_total_patrones(self) -> int:
        """Obtiene total de patrones aprendidos desde DB (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()
                cur.execute('SELECT COUNT(*) FROM aprendizajes_automaticos')
                return cur.fetchone()[0]
        except Exception:
            return 0

    # ---------------------------------------------------------
    # 🔑 FUNCIONES CRÍTICAS DEL CORE
    # ---------------------------------------------------------

    def procesar_pensamiento_cuadruple(self, input_texto: str) -> Dict[str, Any]:
        """PROCESAMIENTO PRINCIPAL - 4 MODALIDADES + APRENDIZAJE."""
        inicio_procesamiento = datetime.now()
        consulta_hash = hashlib.md5(input_texto.encode()).hexdigest()

        # 1. VERIFICAR CACHE
        resultado_cache = self._verificar_cache_consulta_avanzado(consulta_hash)
        if resultado_cache:
            self.metricas["cache_hits"] += 1
            resultado_cache["tiempo_procesamiento"] = (datetime.now() - inicio_procesamiento).total_seconds()
            # Ensure 'cache_utilizado' is in the result
            resultado_cache["metricas_calidad"]["cache_utilizado"] = True
            return resultado_cache

        self.metricas["cache_misses"] += 1

        # 2. ANÁLISIS CUÁDRUPLE (Simulación)
        resultados = {
            "analisis_tecnico": self._modalidad_tecnica_avanzada(input_texto),
            "investigacion_detective": self._modalidad_detective_avanzada(input_texto),
            "exploracion_curiosa": self._modalidad_curiosidad_avanzada(input_texto),
            "analisis_economico": self._modalidad_economica_avanzada(input_texto),
        }

        # 3. PROCESAMIENTO Y SÍNTESIS
        resultados["metricas_calidad"] = self._calcular_metricas_calidad_avanzadas(resultados)
        resultados["sintesis_final"] = self._sintetizar_cuadruple_analisis_optimizada(resultados)

        # 4. APRENDIZAJE Y PERSISTENCIA (Alto Impacto)
        self._procesar_aprendizaje_avanzado(input_texto, resultados)
        self._guardar_analisis_completo(input_texto, consulta_hash, resultados)

        # CALCULAR TIEMPO TOTAL
        resultados["tiempo_procesamiento"] = (datetime.now() - inicio_procesamiento).total_seconds()

        self._actualizar_metricas_globales(resultados) # Actualiza total_analisis

        return resultados

    # ---------------------------------------------------------
    # 💾 MÉTODOS DE PERSISTENCIA (V12.3 - Trazabilidad SQL)
    # ---------------------------------------------------------

    def _guardar_patron_aprendido(self, patron: str, eficacia: float, categoria: str):
        """Guarda y/o actualiza un patrón en DB usando parámetros nombrados (V12.3)."""
        categoria = categoria.lower()
        now = datetime.now().isoformat()

        data_insert = {
            'patron': patron,
            'eficacia': eficacia,
            'uso': 1,
            'ultima_fecha': now,
            'cat': categoria
        }

        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()

                # V12.3: INSERT OR IGNORE con Parámetros Nombrados para claridad
                cursor.execute('''
                    INSERT OR IGNORE INTO aprendizajes_automaticos
                    (patron, eficacia_demostrada, veces_utilizado, ultimo_uso, categoria)
                    VALUES (:patron, :eficacia, :uso, :ultima_fecha, :cat)
                ''', data_insert)

                # V12.3: UPDATE con Parámetros Nombrados
                cursor.execute('''
                    UPDATE aprendizajes_automaticos
                    SET eficacia_demostrada = (eficacia_demostrada * veces_utilizado + :nueva_eficacia) / (veces_utilizado + 1),
                        veces_utilizado = veces_utilizado + 1,
                        ultimo_uso = :ultima_fecha
                    WHERE patron = :patron
                ''', {
                    'nueva_eficacia': eficacia,
                    'ultima_fecha': now,
                    'patron': patron
                })

                conn.commit()

            # Sincronizar el cache en memoria
            self.metricas["patrones_aprendidos"] = self._obtener_total_patrones()
            self.cache_patrones.setdefault(categoria, set()).add(patron)

        except sqlite3.Error as e:
            print(f"❌ ERROR CRÍTICO [SAVE_PATRON]: Fallo de integridad/escritura en DB para patrón '{patron}'. {e}")

    def _guardar_analisis_completo(self, input_texto: str, consulta_hash: str, resultados: Dict[str, Any]):
        """Guarda el registro completo del análisis usando el mapeo fortificado (C2 y V12.3)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()

                # Extracción robusta y dinámica de todos los insights
                all_insights = []
                for modalidad, keys in self.INSIGHT_MAPPING.items():
                    if modalidad in resultados:
                        for key in keys:
                            if isinstance(resultados[modalidad].get(key), list):
                                all_insights.extend(resultados[modalidad][key])

                patrones_detectados = json.dumps(all_insights)
                eficacia = resultados["metricas_calidad"]["eficacia_total"]

                # V12.3: INSERT con Parámetros Nombrados
                cursor.execute('''
                    INSERT INTO analisis_profundos
                    (timestamp, consulta_hash, consulta_texto, eficacia_analisis, patrones_detectados)
                    VALUES (:ts, :hash, :texto, :eficacia, :patrones)
                ''', {
                    'ts': datetime.now().isoformat(),
                    'hash': consulta_hash,
                    'texto': input_texto,
                    'eficacia': eficacia,
                    'patrones': patrones_detectados
                })

                conn.commit()
        except sqlite3.IntegrityError:
            pass # Si el hash ya existe, ignorar
        except sqlite3.Error as e:
            print(f"❌ ERROR CRÍTICO [SAVE_ANALYSIS]: Fallo de escritura en DB para hash '{consulta_hash}'. {e}")

    # ------------------------------------------------------------------
    #  MÉTRICAS DE CALIDAD - COMPLETADO
    # ------------------------------------------------------------------
    def _calcular_metricas_calidad_avanzadas(self, resultados: Dict) -> Dict[str, Any]:
        """Calcula métricas de calidad simuladas y complejas del análisis cuádruple."""
        metricas = {
            "eficacia_total": 0.0,
            "completitud_analisis": 0.0,
            "diversidad_perspectivas": 0.0,
            "utilidad_practica": 0.0,
            "innovacion_propuesta": 0.0,
            "profundidad_tecnica": 0.0,
            "elementos_generados": 0,
            "elementos_utiles": 0
        }

        elementos_totales = 0
        elementos_utiles = 0
        puntuacion_profundidad = 0.0
        peso_eficacia_acumulado = 0.0
        peso_total = 0.0

        for key, analisis in resultados.items():
            if key in ("sintesis_final", "metricas_calidad", "hash_consulta", "timestamp"):
                continue

            # Extraer modalidad (e.g., 'tecnico' de 'analisis_tecnico')
            modalidad = key.split('_')[-1]
            config = self.modalidades_activas.get(modalidad, {"eficacia": 0.0, "peso_analisis": 1.0})
            peso = config["peso_analisis"]
            eficacia_base = config["eficacia"]

            peso_total += peso

            # --- Puntuación basada en los insights generados (simulando profundidad) ---
            for clave, valor in analisis.items():
                if isinstance(valor, list):
                    elementos_totales += len(valor)
                    # Contar listas no vacías como contribución "útil"
                    if len(valor) > 0:
                        elementos_utiles += 1
                        # Contribuir al puntaje de profundidad
                        puntuacion_profundidad += (len(valor) * peso)

            # Contribución de la eficacia ponderada
            peso_eficacia_acumulado += eficacia_base * peso

        # Cálculo de Métricas Finales
        metricas["elementos_generados"] = elementos_totales
        metricas["elementos_utiles"] = elementos_utiles

        if peso_total > 0:
            metricas["eficacia_total"] = peso_eficacia_acumulado / peso_total
        else:
            metricas["eficacia_total"] = 0.0

        # Mapeo del puntaje de profundidad a una métrica legible (máx 1.0)
        metricas["profundidad_tecnica"] = min(puntuacion_profundidad / 15.0, 1.0)

        # Métricas simuladas basadas en eficiencia y profundidad
        metricas["completitud_analisis"] = min(metricas["profundidad_tecnica"] * 0.8 + metricas["eficacia_total"] * 0.2, 1.0)
        metricas["diversidad_perspectivas"] = min(elementos_utiles / 15.0, 1.0)

        # Simular utilidad/innovación basado en el tipo de hallazgo
        if "exploracion_curiosa" in resultados and len(resultados["exploracion_curiosa"].get("horizones_nuevos", [])) > 0:
             metricas["innovacion_propuesta"] = 0.8
        if "analisis_economico" in resultados and len(resultados["analisis_economico"].get("modelo_negocio_sugerido", [])) > 0:
             metricas["utilidad_practica"] = 0.9

        return metricas

    # ------------------------------------------------------------------
    #  MÉTODOS STUB/SINTESIS/UTILITY (Mantenidos de V12.2)
    # ------------------------------------------------------------------

    def _modalidad_tecnica_avanzada(self, texto: str) -> Dict[str, Any]:
        analisis = {"tipo": "TÉCNICO", "patrones_identificados": [], "recomendaciones_implementacion": [], "preguntas_clave": [], "observaciones_tecnicas": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("tecnico", set()) if p in txt]
        if patrones_detectados:
            analisis["patrones_identificados"] = patrones_detectados[:3]
            analisis["recomendaciones_implementacion"].append("Diseñar arquitectura escalable (STUB)")
            analisis["preguntas_clave"].append("¿Qué protocolos criptográficos se aplican?")
            analisis["observaciones_tecnicas"].append(f"🔍 Detectados {len(patrones_detectados)} patrones técnicos.")
        return analisis

    def _modalidad_detective_avanzada(self, texto: str) -> Dict[str, Any]:
        analisis = {"tipo": "DETECTIVE", "pistas_detectadas": [], "riesgos_identificados": [], "hipotesis_iniciales": [], "conexiones_ocultas": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("detective", set()) if p in txt]
        if patrones_detectados:
            analisis["pistas_detectadas"].append("Analizar historial completo de transacciones (STUB)")
            analisis["riesgos_identificados"].append("Riesgo de ataques 51% en chains pequeñas (STUB)")
            analisis["hipotesis_iniciales"].append("Posible relación entre elementos no conectados (STUB)")
            analisis["conexiones_ocultas"].append("Rastrear flujo de fondos entre direcciones (STUB)")
        return analisis

    def _modalidad_curiosidad_avanzada(self, texto: str) -> Dict[str, Any]:
        analisis = {"tipo": "CURIOSIDAD", "preguntas_exploratorias": [], "horizones_nuevos": [], "experimentos_sugeridos": [], "futuros_posibles": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("curiosidad", set()) if p in txt]
        if patrones_detectados:
            analisis["preguntas_exploratorias"].append("¿Qué sucedería si invertimos las premisas actuales? (STUB)")
            analisis["horizones_nuevos"].append("Tendencias emergentes en tecnología blockchain (STUB)")
            analisis["experimentos_sugeridos"].append("Prototipo rápido para validar hipótesis radicales (STUB)")
            analisis["futuros_posibles"].append("Escenario: Adopción masiva de DAOs en 3 años (STUB)")
        return analisis

    def _modalidad_economica_avanzada(self, texto: str) -> Dict[str, Any]:
        analisis = {"tipo": "ECONÓMICO", "metricas_financieras": [], "analisis_rentabilidad": [], "modelo_negocio_sugerido": [], "oportunidades_mercado": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("economico", set()) if p in txt]
        if patrones_detectados:
            analisis["metricas_financieras"].append("Análisis de ROI (Return on Investment) (STUB)")
            analisis["analisis_rentabilidad"].append("Evaluación de retorno sobre inversión (STUB)")
            analisis["modelo_negocio_sugerido"].append("Modelo Freemium con valor añadido en la capa Pro (STUB)")
            analisis["oportunidades_mercado"].append("Identificación de nichos no explotados (STUB)")
        return analisis

    def _verificar_cache_consulta_avanzado(self, consulta_hash: str) -> Optional[Dict[str, Any]]:
        """Busca la respuesta en memoria → DB (C2: Context Manager)."""
        if consulta_hash in self.cache_consultas:
            self.cache_consultas[consulta_hash]["tiempo_procesamiento"] = 0.0001
            return self.cache_consultas[consulta_hash]

        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()
                cursor.execute('SELECT consulta_texto, eficacia_analisis, patrones_detectados FROM analisis_profundos WHERE consulta_hash = :hash ORDER BY timestamp DESC LIMIT 1', {'hash': consulta_hash})
                resultado = cursor.fetchone()

            if resultado:
                _, eficacia, patrones_json = resultado
                patrones_list = json.loads(patrones_json) if patrones_json else []

                resultado_cache = {
                    "sintesis_final": f"⚡ CACHE: Análisis reconstruido. Eficacia previa: {eficacia:.1%}",
                    "metricas_calidad": {"eficacia_total": eficacia * 0.98, "cache_utilizado": True},
                    "tiempo_procesamiento": 0.0001
                }

                if len(self.cache_consultas) >= self.max_cache_size:
                    self.cache_consultas.pop(next(iter(self.cache_consultas)))
                self.cache_consultas[consulta_hash] = resultado_cache

                return resultado_cache

        except sqlite3.Error as e:
            print(f"⚠️ ADVERTENCIA [CACHE_DB_READ]: Error al leer cache desde DB. {e}")

        return None

    def _sintetizar_cuadruple_analisis_optimizada(self, resultados: Dict[str, Any]) -> str:
        """Sintetiza los cuatro análisis en una respuesta coherente."""
        eco = resultados["analisis_economico"]
        tec = resultados["analisis_tecnico"]
        det = resultados["investigacion_detective"]
        cur = resultados["exploracion_curiosa"]
        calidad = resultados["metricas_calidad"]

        sintesis = f"""
**🔍 INFORME DE PENSAMIENTO CUÁDRUPLE (v{self.version})**
**📝 RESUMEN EJECUTIVO (Eficacia: {calidad['eficacia_total']:.1%})**

**🎯 FOCO CLAVE (Prioridad de Ejecución Inmediata):**
1. **Acción Económica:** {eco['modelo_negocio_sugerido'][0] if eco['modelo_negocio_sugerido'] else 'Pendiente de definir modelo.'}
2. **Riesgo Detectado:** {det['riesgos_identificados'][0] if det['riesgos_identificados'] else 'Riesgo bajo o no detectado.'}
3. **Próximo Horizonte:** {cur['horizones_nuevos'][0] if cur['horizones_nuevos'] else 'Exploración de nichos de mercado.'}

**🛠️ DETALLES DE TRAZABILIDAD:**
• Profundidad Técnica: {calidad['profundidad_tecnica']:.1%}
• Diversidad de Perspectivas: {calidad['diversidad_perspectivas']:.1%}
"""
        return sintesis

    def _procesar_aprendizaje_avanzado(self, input_texto: str, resultados: Dict[str, Any]):
        """Aplica lógica para actualizar patrones aprendidos basado en la eficacia."""
        if resultados["metricas_calidad"]["eficacia_total"] > self.umbral_aprendizaje:
            insight_clave = resultados["sintesis_final"].split('\n')[2].strip()
            nuevo_patron = f"Éxito|{insight_clave[:50]}..."
            self._guardar_patron_aprendido(nuevo_patron, resultados["metricas_calidad"]["eficacia_total"], "ÉXITO")
            self.metricas["mejoras_detectadas"] += 1

    def _actualizar_metricas_globales(self, resultados: Dict[str, Any]):
        """Actualiza métricas tras cada análisis."""
        self.metricas["total_analisis"] += 1
        nueva_eficacia = resultados["metricas_calidad"]["eficacia_total"]
        alpha = 0.1
        self.metricas["eficiencia_promedio"] = (
            (1 - alpha) * self.metricas["eficiencia_promedio"] + alpha * nueva_eficacia
        )
        # Se omitió la lógica de guardado periódico en este archivo, enfocándose en el Core.

    def obtener_estado_sistema(self) -> str:
        return f"""
🧠 {self.nombre} - ESTADO DEL SISTEMA (v{self.version})
{'=' * 50}

📊 MÉTRICAS PRINCIPALES
• Análisis realizados         : {self.metricas['total_analisis']:,}
• Eficiencia promedio       : {self.metricas['eficiencia_promedio']:.1%}
• Patrones aprendidos       : {self.metricas['patrones_aprendidos']}
• Cache hits / misses       : {self.metricas['cache_hits']} / {self.metricas['cache_misses']}
"""

# =========================================================
# 🚀 INTERFAZ DE PRUEBA DEL SISTEMA CUÁDRUPLE
# =========================================================

def ejecutar_core_alto_impacto():
    """Ejecuta la interfaz de prueba para el Core de Alto Impacto."""

    # 🧹 Limpieza de archivos de prueba
    if os.path.exists("vida_core_evolucionado.db"):
         os.remove("vida_core_evolucionado.db")

    vida = VidaCoreEvolucionado()

    print("\n" + "=" * 70)
    print(f"🧠 {vida.nombre} - MODO DESPLIEGUE INMEDIATO v{vida.version}")
    print("=" * 70)

    # Simulación de un proceso de aprendizaje inicial para que el cache se actualice
    print("🎓 Pre-carga de conocimiento (Simulando 2 aprendizajes iniciales)...")
    vida._guardar_patron_aprendido("Optimización ROI cripto", 0.95, "ECONOMICO")
    vida._guardar_patron_aprendido("Defensa contra reentrancy", 0.98, "TECNICO")
    print(f"Patrones en DB: {vida.metricas['patrones_aprendidos']}")

    consulta_1 = "Analizar la viabilidad económica de un modelo de staking con alto riesgo técnico."

    # Primera consulta (forzará un cache miss y la persistencia)
    print(f"\n[1/2] CONSULTA CRÍTICA: '{consulta_1}'")
    resultado_1 = vida.procesar_pensamiento_cuadruple(consulta_1)
    print(resultado_1["sintesis_final"])
    # Check if 'cache_utilizado' exists before accessing it
    cache_status_1 = resultado_1.get('metricas_calidad', {}).get('cache_utilizado', False)
    print(f"⏱️ Tiempo: {resultado_1.get('tiempo_procesamiento', 0.0):.4f}s | Cache: {cache_status_1}")


    # Segunda consulta (debería ser un cache hit)
    print("\n[2/2] CONSULTA REPETIDA (Verificando Cache)...")
    time.sleep(0.01) # Pequeña pausa
    resultado_2 = vida.procesar_pensamiento_cuadruple(consulta_1)
    print(resultado_2["sintesis_final"])
    # Check if 'cache_utilizado' exists before accessing it
    cache_status_2 = resultado_2.get('metricas_calidad', {}).get('cache_utilizado', False)
    print(f"⏱️ Tiempo: {resultado_2.get('tiempo_procesamiento', 0.0):.4f}s | Cache: {cache_status_2}")


    print("\n" + "=" * 70)
    print(f"📊 ESTADO FINAL DEL SISTEMA V{vida.version}")
    print(vida.obtener_estado_sistema())
    print("=" * 70)

if __name__ == "__main__":
    ejecutar_core_alto_impacto()
```

```
class VidaCoreEvolucionado:
    """
    🚀 SISTEMA DE PENSAMIENTO CUÁDRUPLE EVOLUCIONADO
    Versión 12.3 - Fortificación de Arquitectura de Producción y Trazabilidad.
    """

    # Mapeo para garantizar la extracción completa de insights (R1 Fortificada)
    INSIGHT_MAPPING = {
        "analisis_tecnico": ["patrones_identificados", "recomendaciones_implementacion", "preguntas_clave", "observaciones_tecnicas"],
        "investigacion_detective": ["pistas_detectadas", "riesgos_identificados", "hipotesis_iniciales", "conexiones_ocultas"],
        "exploracion_curiosa": ["preguntas_exploratorias", "horizones_nuevos", "experimentos_sugeridos", "futuros_posibles"],
        "analisis_economico": ["metricas_financieras", "modelo_negocio_sugerido", "analisis_rentabilidad", "oportunidades_mercado"]
    }

    def __init__(self, nombre: str = "VIDA-CORE-EVOLUCIONADO"):
        self.nombre = nombre
        self.creador = "Daniel"
        self.version = "12.3 - Máxima Trazabilidad"
        self.estado = "PENSAMIENTO_ACTIVO"
        self.base_datos = "vida_core_evolucionado.db"
        self.umbral_aprendizaje = 0.7

        # 🎯 ARQUITECTURA CUÁDRUPLE MEJORADA
        self.modalidades_activas = {
            "tecnico": {"nivel": 10, "eficacia": 0.92, "peso_analisis": 1.2, "patrones": ["código", "software", "algoritmo", "sistema", "seguridad", "blockchain", "contrato", "api", "vulnerability", "smart contract"]},
            "detective": {"nivel": 10, "eficacia": 0.88, "peso_analisis": 1.1, "patrones": ["investigar", "patrón", "vulnerabilidad", "forense", "auditoría", "transacción", "conexión", "unusual transaction", "large transfer", "exploit", "rug pull", "exit scam"]},
            "curiosidad": {"nivel": 10, "eficacia": 0.85, "peso_analisis": 1.0, "patrones": ["qué pasaría si", "innovación", "futuro", "disrupción", "explorar", "nuevo", "optimizar", "what if", "how could", "future of", "emerging trends", "experiment", "prototype", "potential", "could be"]},
            "economico": {"nivel": 9, "eficacia": 0.80, "peso_analisis": 0.9, "patrones": ["costo", "inversión", "rentabilidad", "ROI", "mercado", "negocio", "modelo", "financiero", "return on investment", "business model", "revenue stream", "market opportunity", "niche", "fee", "freemium"]}
        }

        # 🚀 SISTEMA DE PERSISTENCIA AVANZADO
        self._inicializar_base_datos_avanzada()

        # ⚡ CACHE DE ALTO RENDIMIENTO
        self.cache_patrones = self._construir_cache_patrones()
        self.cache_consultas = {}
        self.max_cache_size = 1000

        # 📊 MÉTRICAS Y ESTADO
        self.metricas = {
            "total_analisis": self._obtener_total_analisis(),
            "eficiencia_promedio": self._obtener_eficiencia_promedio(),
            "patrones_aprendidos": self._obtener_total_patrones(),
            "mejoras_detectadas": 0,
            "cache_hits": 0,
            "cache_misses": 0
        }

        print(f"🚀 {self.nombre} INICIALIZADO - v{self.version}")
        print(f"📊 Base: {self.metricas['total_analisis']} análisis | {self.metricas['patrones_aprendidos']} patrones persistentes.")

    # ------------------------------------------------------------------
    #  Métodos de bajo nivel (DB, caché, métricas...) - CORREGIDOS (C2)
    # ------------------------------------------------------------------

    def _inicializar_base_datos_avanzada(self):
        """Crea tablas e índices si no existen (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()

                # Tabla principal de análisis
                cur.execute('''
                    CREATE TABLE IF NOT EXISTS analisis_profundos (
                        id INTEGER PRIMARY KEY AUTOINCREMENT,
                        timestamp TEXT NOT NULL,
                        consulta_hash TEXT UNIQUE,
                        consulta_texto TEXT NOT NULL,
                        eficacia_analisis REAL,
                        patrones_detectados TEXT
                    )
                ''')

                # Aprendizajes automáticos
                cur.execute('''
                    CREATE TABLE IF NOT EXISTS aprendizajes_automaticos (
                        id INTEGER PRIMARY KEY AUTOINCREMENT,
                        patron TEXT UNIQUE,
                        eficacia_demostrada REAL DEFAULT 0.7,
                        veces_utilizado INTEGER DEFAULT 1,
                        ultimo_uso TEXT,
                        categoria TEXT
                    )
                ''')

                cur.execute('CREATE INDEX IF NOT EXISTS idx_hash ON analisis_profundos(consulta_hash)')
                cur.execute('CREATE INDEX IF NOT EXISTS idx_patron ON aprendizajes_automaticos(patron)')
                conn.commit()
        except Exception as e:
            print(f"❌ ERROR CRÍTICO [INIT_DB]: No se pudo inicializar la base de datos. {e}")

    def _construir_cache_patrones(self) -> Dict[str, set]:
        """Carga patrones fijos y aprendidos en el cache de memoria (C2: Context Manager)."""
        cache = {}
        # Cargar patrones fijos de las modalidades
        for modalidad, datos in self.modalidades_activas.items():
            cache[modalidad] = set(datos['patrones'])

        # Cargar patrones aprendidos desde la DB
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()
                cursor.execute("SELECT patron, categoria FROM aprendizajes_automaticos")
                for patron, categoria in cursor.fetchall():
                    cache.setdefault(categoria.lower(), set()).add(patron)
        except Exception:
            pass # Si falla, solo usamos cache fijo

        return cache

    def _obtener_total_analisis(self) -> int:
        """Obtiene el total de análisis desde la DB (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()
                cur.execute('SELECT COUNT(*) FROM analisis_profundos')
                return cur.fetchone()[0]
        except Exception:
            return 0

    def _obtener_eficiencia_promedio(self) -> float:
        """Calcula eficiencia promedio desde DB (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()
                cur.execute('SELECT AVG(eficacia_analisis) FROM analisis_profundos WHERE eficacia_analisis > 0')
                resultado = cur.fetchone()[0]
                return resultado if resultado else 0.75
        except Exception:
            return 0.75

    def _obtener_total_patrones(self) -> int:
        """Obtiene total de patrones aprendidos desde DB (C2: Context Manager)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cur = conn.cursor()
                cur.execute('SELECT COUNT(*) FROM aprendizajes_automaticos')
                return cur.fetchone()[0]
        except Exception:
            return 0

    # ---------------------------------------------------------
    # 🔑 FUNCIONES CRÍTICAS DEL CORE
    # ---------------------------------------------------------

    def procesar_pensamiento_cuadruple(self, input_texto: str) -> Dict[str, Any]:
        """PROCESAMIENTO PRINCIPAL - 4 MODALIDADES + APRENDIZAJE."""
        inicio_procesamiento = datetime.now()
        consulta_hash = hashlib.md5(input_texto.encode()).hexdigest()

        # 1. VERIFICAR CACHE
        resultado_cache = self._verificar_cache_consulta_avanzado(consulta_hash)
        if resultado_cache:
            self.metricas["cache_hits"] += 1
            resultado_cache["tiempo_procesamiento"] = (datetime.now() - inicio_procesamiento).total_seconds()
            # Ensure 'cache_utilizado' is in the result
            resultado_cache["metricas_calidad"]["cache_utilizado"] = True
            return resultado_cache

        self.metricas["cache_misses"] += 1

        # 2. ANÁLISIS CUÁDRUPLE (Simulación)
        resultados = {
            "analisis_tecnico": self._modalidad_tecnica_avanzada(input_texto),
            "investigacion_detective": self._modalidad_detective_avanzada(input_texto),
            "exploracion_curiosa": self._modalidad_curiosidad_avanzada(input_texto),
            "analisis_economico": self._modalidad_economica_avanzada(input_texto),
        }

        # 3. PROCESAMIENTO Y SÍNTESIS
        resultados["metricas_calidad"] = self._calcular_metricas_calidad_avanzadas(resultados)
        resultados["sintesis_final"] = self._sintetizar_cuadruple_analisis_optimizada(resultados)

        # 4. APRENDIZAJE Y PERSISTENCIA (Alto Impacto)
        self._procesar_aprendizaje_avanzado(input_texto, resultados)
        self._guardar_analisis_completo(input_texto, consulta_hash, resultados)

        # CALCULAR TIEMPO TOTAL
        resultados["tiempo_procesamiento"] = (datetime.now() - inicio_procesamiento).total_seconds()

        self._actualizar_metricas_globales(resultados) # Actualiza total_analisis

        return resultados

    # ---------------------------------------------------------
    # 💾 MÉTODOS DE PERSISTENCIA (V12.3 - Trazabilidad SQL)
    # ---------------------------------------------------------

    def _guardar_patron_aprendido(self, patron: str, eficacia: float, categoria: str):
        """Guarda y/o actualiza un patrón en DB usando parámetros nombrados (V12.3)."""
        categoria = categoria.lower()
        now = datetime.now().isoformat()

        data_insert = {
            'patron': patron,
            'eficacia': eficacia,
            'uso': 1,
            'ultima_fecha': now,
            'cat': categoria
        }

        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()

                # V12.3: INSERT OR IGNORE con Parámetros Nombrados para claridad
                cursor.execute('''
                    INSERT OR IGNORE INTO aprendizajes_automaticos
                    (patron, eficacia_demostrada, veces_utilizado, ultimo_uso, categoria)
                    VALUES (:patron, :eficacia, :uso, :ultima_fecha, :cat)
                ''', data_insert)

                # V12.3: UPDATE con Parámetros Nombrados
                cursor.execute('''
                    UPDATE aprendizajes_automaticos
                    SET eficacia_demostrada = (eficacia_demostrada * veces_utilizado + :nueva_eficacia) / (veces_utilizado + 1),
                        veces_utilizado = veces_utilizado + 1,
                        ultimo_uso = :ultima_fecha
                    WHERE patron = :patron
                ''', {
                    'nueva_eficacia': eficacia,
                    'ultima_fecha': now,
                    'patron': patron
                })

                conn.commit()

            # Sincronizar el cache en memoria
            self.metricas["patrones_aprendidos"] = self._obtener_total_patrones()
            self.cache_patrones.setdefault(categoria, set()).add(patron)

        except sqlite3.Error as e:
            print(f"❌ ERROR CRÍTICO [SAVE_PATRON]: Fallo de integridad/escritura en DB para patrón '{patron}'. {e}")

    def _guardar_analisis_completo(self, input_texto: str, consulta_hash: str, resultados: Dict[str, Any]):
        """Guarda el registro completo del análisis usando el mapeo fortificado (C2 y V12.3)."""
        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()

                # Extracción robusta y dinámica de todos los insights
                all_insights = []
                for modalidad, keys in self.INSIGHT_MAPPING.items():
                    if modalidad in resultados:
                        for key in keys:
                            if isinstance(resultados[modalidad].get(key), list):
                                all_insights.extend(resultados[modalidad][key])

                patrones_detectados = json.dumps(all_insights)
                eficacia = resultados["metricas_calidad"]["eficacia_total"]

                # V12.3: INSERT con Parámetros Nombrados
                cursor.execute('''
                    INSERT INTO analisis_profundos
                    (timestamp, consulta_hash, consulta_texto, eficacia_analisis, patrones_detectados)
                    VALUES (:ts, :hash, :texto, :eficacia, :patrones)
                ''', {
                    'ts': datetime.now().isoformat(),
                    'hash': consulta_hash,
                    'texto': input_texto,
                    'eficacia': eficacia,
                    'patrones': patrones_detectados
                })

                conn.commit()
        except sqlite3.IntegrityError:
            pass # Si el hash ya existe, ignorar
        except sqlite3.Error as e:
            print(f"❌ ERROR CRÍTICO [SAVE_ANALYSIS]: Fallo de escritura en DB para hash '{consulta_hash}'. {e}")

    # ------------------------------------------------------------------
    #  MÉTRICAS DE CALIDAD - COMPLETADO
    # ------------------------------------------------------------------
    def _calcular_metricas_calidad_avanzadas(self, resultados: Dict) -> Dict[str, Any]:
        """Calcula métricas de calidad simuladas y complejas del análisis cuádruple."""
        metricas = {
            "eficacia_total": 0.0,
            "completitud_analisis": 0.0,
            "diversidad_perspectivas": 0.0,
            "utilidad_practica": 0.0,
            "innovacion_propuesta": 0.0,
            "profundidad_tecnica": 0.0,
            "elementos_generados": 0,
            "elementos_utiles": 0,
            "cache_utilizado": False # Add cache_utilizado here
        }

        elementos_totales = 0
        elementos_utiles = 0
        puntuacion_profundidad = 0.0
        peso_eficacia_acumulado = 0.0
        peso_total = 0.0

        for key, analisis in resultados.items():
            if key in ("sintesis_final", "metricas_calidad", "hash_consulta", "timestamp"):
                continue

            # Extraer modalidad (e.g., 'tecnico' de 'analisis_tecnico')
            modalidad = key.split('_')[-1]
            config = self.modalidades_activas.get(modalidad, {"eficacia": 0.0, "peso_analisis": 1.0})
            peso = config["peso_analisis"]
            eficacia_base = config["eficacia"]

            peso_total += peso

            # --- Puntuación basada en los insights generados (simulando profundidad) ---
            for clave, valor in analisis.items():
                if isinstance(valor, list):
                    elementos_totales += len(valor)
                    # Contar listas no vacías como contribución "útil"
                    if len(valor) > 0:
                        elementos_utiles += 1
                        # Contribuir al puntaje de profundidad
                        puntuacion_profundidad += (len(valor) * peso)

            # Contribución de la eficacia ponderada
            peso_eficacia_acumulado += eficacia_base * peso

        # Cálculo de Métricas Finales
        metricas["elementos_generados"] = elementos_totales
        metricas["elementos_utiles"] = elementos_utiles

        if peso_total > 0:
            metricas["eficacia_total"] = peso_eficacia_acumulado / peso_total
        else:
            metricas["eficacia_total"] = 0.0

        # Mapeo del puntaje de profundidad a una métrica legible (máx 1.0)
        metricas["profundidad_tecnica"] = min(puntuacion_profundidad / 15.0, 1.0)

        # Métricas simuladas basadas en eficiencia y profundidad
        metricas["completitud_analisis"] = min(metricas["profundidad_tecnica"] * 0.8 + metricas["eficacia_total"] * 0.2, 1.0)
        metricas["diversidad_perspectivas"] = min(elementos_utiles / 15.0, 1.0)

        # Simular utilidad/innovación basado en el tipo de hallazgo
        if "exploracion_curiosa" in resultados and len(resultados["exploracion_curiosa"].get("horizones_nuevos", [])) > 0:
             metricas["innovacion_propuesta"] = 0.8
        if "analisis_economico" in resultados and len(resultados["analisis_economico"].get("modelo_negocio_sugerido", [])) > 0:
             metricas["utilidad_practica"] = 0.9

        # If cache was used, metrics are different (simulated)
        if resultados.get("metricas_calidad", {}).get("cache_utilizado", False):
             # These values are set during cache hit, just ensure they are consistent
             pass


        return metricas

    # ------------------------------------------------------------------
    #  MÉTODOS STUB/SINTESIS/UTILITY (Mantenidos de V12.2)
    # ------------------------------------------------------------------

    def _modalidad_tecnica_avanzada(self, texto: str) -> Dict[str, Any]:
        """Advanced Technical Analysis: Identifies technical patterns and provides insights."""
        analisis = {"tipo": "TÉCNICO", "patrones_identificados": [], "recomendaciones_implementacion": [], "preguntas_clave": [], "observaciones_tecnicas": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("tecnico", set()) if p in txt]

        if patrones_detectados:
            # Implement more sophisticated logic here (simulated)
            analisis["patrones_identificados"] = list(set(patrones_detectados))[:5] # Use set to avoid duplicates

            if "smart contract" in txt and "vulnerability" in txt:
                analisis["recomendaciones_implementacion"].append("Realizar auditoría de seguridad exhaustiva del contrato.")
                analisis["riesgos_identificados"] = ["Vulnerabilidad potencial en contrato inteligente."] # Added to detective
                analisis["preguntas_clave"].append("¿Se han realizado pruebas de fuzzing en el contrato?")
                analisis["observaciones_tecnicas"].append("Se detectaron términos relacionados con vulnerabilidades en contratos inteligentes.")
            elif "blockchain" in txt and "consensus" in txt:
                analisis["recomendaciones_implementacion"].append("Evaluar el mecanismo de consenso para escalabilidad.")
                analisis["preguntas_clave"].append("¿Cuál es la latencia de confirmación de transacciones?")
                analisis["observaciones_tecnicas"].append("Análisis enfocado en el mecanismo de consenso de la blockchain.")
            elif "api" in txt and "security" in txt:
                 analisis["recomendaciones_implementacion"].append("Implementar autenticación y autorización robustas en la API.")
                 analisis["preguntas_clave"].append("¿Se utilizan tokens de acceso con expiración?")
                 analisis["observaciones_tecnicas"].append("Se identificaron preocupaciones de seguridad relacionadas con la API.")

            if len(patrones_detectados) > 5:
                 analisis["observaciones_tecnicas"].append(f"Se identificaron un total de {len(patrones_detectados)} patrones técnicos.")

        return analisis

    def _modalidad_detective_avanzada(self, texto: str) -> Dict[str, Any]:
        """Advanced Detective Analysis: Identifies suspicious patterns and potential risks."""
        analisis = {"tipo": "DETECTIVE", "pistas_detectadas": [], "riesgos_identificados": [], "hipotesis_iniciales": [], "conexiones_ocultas": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("detective", set()) if p in txt]

        if patrones_detectados:
            # Implement more sophisticated logic here (simulated)
            if "unusual transaction" in txt or "large transfer" in txt:
                analisis["pistas_detectadas"].append("Actividad transaccional inusual detectada.")
                analisis["riesgos_identificados"].append("Riesgo de manipulación o actividad ilícita.")
                analisis["hipotesis_iniciales"].append("Posible intento de lavado de dinero o manipulación de mercado.")
                analisis["conexiones_ocultas"].append("Investigar conexiones entre direcciones de origen y destino.")
            if "vulnerability" in txt or "exploit" in txt:
                 analisis["pistas_detectadas"].append("Términos relacionados con exploits o vulnerabilidades detectados.")
                 analisis["riesgos_identificados"].append("Riesgo de ataque inminente o pasado.")
                 analisis["hipotesis_iniciales"].append("Posible vector de ataque identificado.")
            if "rug pull" in txt or "exit scam" in txt:
                 analisis["pistas_detectadas"].append("Advertencias sobre posibles estafas detectadas.")
                 analisis["riesgos_identificados"].append("Alto riesgo de pérdida de inversión.")
                 analisis["hipotesis_iniciales"].append("Posible actividad fraudulenta planeada.")

            if len(patrones_detectados) > 0:
                 analisis["observaciones_tecnicas"] = [f"Se identificaron {len(patrones_detectados)} patrones con indicios sospechosos."] # Reusing observations_tecnicas for detective

        return analisis


    def _modalidad_curiosidad_avanzada(self, texto: str) -> Dict[str, Any]:
        """Advanced Curiosity Analysis: Explores forward-looking and speculative ideas."""
        analisis = {"tipo": "CURIOSIDAD", "preguntas_exploratorias": [], "horizones_nuevos": [], "experimentos_sugeridos": [], "futuros_posibles": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("curiosidad", set()) if p in txt]

        if patrones_detectados:
            # Implement more sophisticated logic here (simulated)
            if "what if" in txt or "how could" in txt:
                analisis["preguntas_exploratorias"].append("Explorar escenarios hipotéticos y sus implicaciones.")
            if "future of" in txt or "emerging trends" in txt:
                analisis["horizones_nuevos"].append("Identificar tendencias y áreas de innovación emergente.")
            if "experiment" in txt or "prototype" in txt:
                 analisis["experimentos_sugeridos"].append("Proponer experimentos o pruebas de concepto.")
            if "potential" in txt or "could be" in txt:
                 analisis["futuros_posibles"].append("Visualizar posibles escenarios futuros basados en la información.")

            if len(patrones_detectados) > 0:
                 analisis["observaciones_tecnicas"] = [f"Se identificaron {len(patrones_detectados)} patrones con potencial exploratorio."] # Reusing observations_tecnicas

        return analisis

    def _modalidad_economica_avanzada(self, texto: str) -> Dict[str, Any]:
        """Advanced Economic Analysis: Analyzes financial aspects, models, and market trends."""
        analisis = {"tipo": "ECONÓMICO", "metricas_financieras": [], "analisis_rentabilidad": [], "modelo_negocio_sugerido": [], "oportunidades_mercado": []}
        txt = texto.lower()
        patrones_detectados = [p for p in self.cache_patrones.get("economico", set()) if p in txt]

        if patrones_detectados:
            # Implement more sophisticated logic here (simulated)
            if "roi" in txt or "return on investment" in txt:
                analisis["metricas_financieras"].append("Análisis de Retorno de la Inversión (ROI).")
                analisis["analisis_rentabilidad"].append("Evaluar la rentabilidad esperada de la iniciativa.")
            if "business model" in txt or "revenue stream" in txt:
                analisis["modelo_negocio_sugerido"].append("Sugerir modelos de negocio o fuentes de ingresos.")
            if "market opportunity" in txt or "niche" in txt:
                 analisis["oportunidades_mercado"].append("Identificar oportunidades de mercado o nichos sin explotar.")
            if "cost" in txt or "fee" in txt:
                 analisis["metricas_financieras"].append("Considerar costos y tarifas asociadas.")
            if "freemium" in txt:
                 analisis["modelo_negocio_sugerido"].append("Considerar modelo Freemium con valor añadido.")


            if len(patrones_detectados) > 0:
                 analisis["observaciones_tecnicas"] = [f"Se identificaron {len(patrones_detectados)} patrones con relevancia económica."] # Reusing observations_tecnicas

        return analisis

    def _verificar_cache_consulta_avanzado(self, consulta_hash: str) -> Optional[Dict[str, Any]]:
        """Busca la respuesta en memoria → DB (C2: Context Manager)."""
        if consulta_hash in self.cache_consultas:
            self.cache_consultas[consulta_hash]["tiempo_procesamiento"] = 0.0001
            return self.cache_consultas[consulta_hash]

        try:
            with sqlite3.connect(self.base_datos) as conn:
                cursor = conn.cursor()
                cursor.execute('SELECT consulta_texto, eficacia_analisis, patrones_detectados FROM analisis_profundos WHERE consulta_hash = :hash ORDER BY timestamp DESC LIMIT 1', {'hash': consulta_hash})
                resultado = cursor.fetchone()

            if resultado:
                _, eficacia, patrones_json = resultado
                patrones_list = json.loads(patrones_json) if patrones_json else []

                # Reconstruct a simplified result for cache hit
                resultado_cache = {
                    "sintesis_final": f"⚡ CACHE: Análisis reconstruido. Eficacia previa: {eficacia:.1%}",
                    "metricas_calidad": {"eficacia_total": eficacia, "cache_utilizado": True},
                    "tiempo_procesamiento": 0.0001,
                    # Add empty lists for expected keys to avoid errors later
                    "analisis_tecnico": {"patrones_identificados": [], "recomendaciones_implementacion": [], "preguntas_clave": [], "observaciones_tecnicas": []},
                    "investigacion_detective": {"pistas_detectadas": [], "riesgos_identificados": [], "hipotesis_iniciales": [], "conexiones_ocultas": []},
                    "exploracion_curiosa": {"preguntas_exploratorias": [], "horizones_nuevos": [], "experimentos_sugeridos": [], "futuros_posibles": []},
                    "analisis_economico": {"metricas_financieras": [], "modelo_negocio_sugerido": [], "analisis_rentabilidad": [], "oportunidades_mercado": []},
                }

                # Attempt to populate with patterns if available (optional)
                if patrones_list:
                     # Simple distribution based on keywords (this is a simulation)
                     for pattern in patrones_list:
                         if any(kw in pattern.lower() for kw in ["contrato", "technical", "seguridad"]):
                            resultado_cache["analisis_tecnico"]["patrones_identificados"].append(pattern)
                         elif any(kw in pattern.lower() for kw in ["riesgo", "suspic", "detective"]):
                             resultado_cache["investigacion_detective"]["pistas_detectadas"].append(pattern)
                         elif any(kw in pattern.lower() for kw in ["future", "explor", "curiosidad"]):
                             resultado_cache["exploracion_curiosa"]["horizones_nuevos"].append(pattern)
                         elif any(kw in pattern.lower() for kw in ["economic", "mercado", "financiero"]):
                             resultado_cache["analisis_economico"]["metricas_financieras"].append(pattern)


                if len(self.cache_consultas) >= self.max_cache_size:
                    self.cache_consultas.pop(next(iter(self.cache_consultas)))
                self.cache_consultas[consulta_hash] = resultado_cache

                return resultado_cache

        except sqlite3.Error as e:
            print(f"⚠️ ADVERTENCIA [CACHE_DB_READ]: Error al leer cache desde DB. {e}")

        return None


    def _sintetizar_cuadruple_analisis_optimizada(self, resultados: Dict[str, Any]) -> str:
        """Sintetiza los cuatro análisis en una respuesta coherente, handling empty lists."""
        eco = resultados.get("analisis_economico", {})
        tec = resultados.get("analisis_tecnico", {})
        det = resultados.get("investigacion_detective", {})
        cur = resultados.get("exploracion_curiosa", {})
        calidad = resultados.get("metricas_calidad", {})

        # Helper function to get the first element or a default message
        def get_first_or_default(lst, default_msg):
            return lst[0] if lst else default_msg

        sintesis = f"""
**🔍 INFORME DE PENSAMIENTO CUÁDRUPLE (v{self.version})**
**📝 RESUMEN EJECUTIVO (Eficacia: {calidad.get('eficacia_total', 0.0):.1%})**

**🎯 FOCO CLAVE (Prioridad de Ejecución Inmediata):**
1. **Acción Económica:** {get_first_or_default(eco.get('modelo_negocio_sugerido'), 'Pendiente de definir modelo.')}
2. **Riesgo Detectado:** {get_first_or_default(det.get('riesgos_identificados'), 'Riesgo bajo o no detectado.')}
3. **Próximo Horizonte:** {get_first_or_default(cur.get('horizones_nuevos'), 'Exploración de nuevos horizontes.')}

**🛠️ DETALLES DE TRAZABILIDAD:**
• Profundidad Técnica: {calidad.get('profundidad_tecnica', 0.0):.1%}
• Diversidad de Perspectivas: {calidad.get('diversidad_perspectivas', 0.0):.1%}
• Cache Utilizado: {calidad.get('cache_utilizado', False)}
"""
        return sintesis


    def _procesar_aprendizaje_avanzado(self, input_texto: str, resultados: Dict[str, Any]):
        """Aplica lógica para actualizar patrones aprendidos basado en la eficacia."""
        if resultados["metricas_calidad"]["eficacia_total"] > self.umbral_aprendizaje:
            # Extract key insight from relevant modalities based on INSIGHT_MAPPING
            key_insights = []
            for modality, keys in self.INSIGHT_MAPPING.items():
                if modality in resultados:
                    for key in keys:
                        if isinstance(resultados[modality].get(key), list) and resultados[modality].get(key):
                            # Take the first item from the list as a potential key insight
                            key_insights.append(resultados[modality][key][0])

            if key_insights:
                # Use a combination or the first key insight as the new pattern
                # Limit pattern length to avoid overly long patterns
                nuevo_patron = f"Éxito|{' - '.join(key_insights)[:100]}..."
                self._guardar_patron_aprendido(nuevo_patron, resultados["metricas_calidad"]["eficacia_total"], "ÉXITO")
                self.metricas["mejoras_detectadas"] += 1


    def _actualizar_metricas_globales(self, resultados: Dict[str, Any]):
        """Actualiza métricas tras cada análisis."""
        self.metricas["total_analisis"] += 1
        nueva_eficacia = resultados["metricas_calidad"]["eficacia_total"]
        alpha = 0.1
        self.metricas["eficiencia_promedio"] = (
            (1 - alpha) * self.metricas["eficiencia_promedio"] + alpha * nueva_eficacia
        )
        # Se omitió la lógica de guardado periódico en este archivo, enfocándose en el Core.

    def obtener_estado_sistema(self) -> str:
        return f"""
🧠 {self.nombre} - ESTADO DEL SISTEMA (v{self.version})
{'=' * 50}

📊 MÉTRICAS PRINCIPALES
• Análisis realizados         : {self.metricas['total_analisis']:,}
• Eficiencia promedio       : {self.metricas['eficiencia_promedio']:.1%}
• Patrones aprendidos       : {self.metricas['patrones_aprendidos']}
• Cache hits / misses       : {self.metricas['cache_hits']} / {self.metricas['cache_misses']}
"""

# =========================================================
# 🚀 INTERFAZ DE PRUEBA DEL SISTEMA CUÁDRUPLE
# =========================================================

def ejecutar_core_alto_impacto():
    """Ejecuta la interfaz de prueba para el Core de Alto Impacto."""

    # 🧹 Limpieza de archivos de prueba
    if os.path.exists("vida_core_evolucionado.db"):
         os.remove("vida_core_evolucionado.db")

    vida = VidaCoreEvolucionado()

    print("\n" + "=" * 70)
    print(f"🧠 {vida.nombre} - MODO DESPLIEGUE INMEDIATO v{vida.version}")
    print("=" * 70)

    # Simulación de un proceso de aprendizaje inicial para que el cache se actualice
    print("🎓 Pre-carga de conocimiento (Simulando 2 aprendizajes iniciales)...")
    vida._guardar_patron_aprendido("Optimización ROI cripto", 0.95, "ECONOMICO")
    vida._guardar_patron_aprendido("Defensa contra reentrancy", 0.98, "TECNICO")
    print(f"Patrones en DB: {vida.metricas['patrones_aprendidos']}")

    consulta_1 = "Analizar la viabilidad económica de un modelo de staking con alto riesgo técnico, considerando posibles vulnerabilidades en smart contracts y oportunidades de mercado en nichos emergentes. ¿Qué sucedería si implementamos un modelo freemium?"

    # Primera consulta (forzará un cache miss y la persistencia)
    print(f"\n[1/2] CONSULTA CRÍTICA: '{consulta_1}'")
    resultado_1 = vida.procesar_pensamiento_cuadruple(consulta_1)
    print(resultado_1["sintesis_final"])
    # Check if 'cache_utilizado' exists before accessing it
    cache_status_1 = resultado_1.get('metricas_calidad', {}).get('cache_utilizado', False)
    print(f"⏱️ Tiempo: {resultado_1.get('tiempo_procesamiento', 0.0):.4f}s | Cache: {cache_status_1}")


    # Segunda consulta (debería ser un cache hit)
    print("\n[2/2] CONSULTA REPETIDA (Verificando Cache)...")
    time.sleep(0.01) # Pequeña pausa
    resultado_2 = vida.procesar_pensamiento_cuadruple(consulta_1)
    print(resultado_2["sintesis_final"])
    # Check if 'cache_utilizado' exists before accessing it
    cache_status_2 = resultado_2.get('metricas_calidad', {}).get('cache_utilizado', False)
    print(f"⏱️ Tiempo: {resultado_2.get('tiempo_procesamiento', 0.0):.4f}s | Cache: {cache_status_2}")


    print("\n" + "=" * 70)
    print(f"📊 ESTADO FINAL DEL SISTEMA V{vida.version}")
    print(vida.obtener_estado_sistema())
    print("=" * 70)

if __name__ == "__main__":
    ejecutar_core_alto_impacto()
```
